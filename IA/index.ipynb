```python
def predecir_dificultad(edad, peso, objetivo, dias, tiempo, experiencia, lugar, lesiones):
    usuario = pd.DataFrame({
        "edad": [edad],
        "peso_kg": [peso],
        "objetivo": [objetivo],
        "dias_disponibles": [dias],
        "tiempo_disponible": [tiempo],
        "experiencia": [experiencia],
        "lugar_entrenamiento": [lugar],
        "lesiones": [lesiones]
    })

    usuario_convertido = pd.get_dummies(usuario)

    usuario_convertido = usuario_convertido.reindex(
        columns=X_convertido.columns,
        fill_value=False
    )

    prediccion = modelo.predict(usuario_convertido)[0]

    dificultades = {
        0: "Facil",
        1: "Medio",
        2: "Dificil"
    }

    return dificultades[prediccion]
```

```python
def evaluar_usuario(
    edad,
    peso,
    objetivo,
    dias,
    tiempo,
    experiencia,
    lugar,
    zona_lesion,
    estado_lesion
):
    dificultad = predecir_dificultad(
        edad,
        peso,
        objetivo,
        dias,
        tiempo,
        experiencia,
        lugar,
        zona_lesion if zona_lesion != "nada" else "nada"
    )

    if estado_lesion == "actual":
        tipo_adaptacion = "especial"
        accion = f"limitar ejercicios de {zona_lesion} y requerir revision profesional"

    elif estado_lesion == "pasada":
        tipo_adaptacion = "preventiva"
        accion = f"adaptar ejercicios para cuidar {zona_lesion}"

    else:
        tipo_adaptacion = "ninguna"
        accion = "rutina normal"

    return {
        "dificultad": dificultad,
        "tipo_adaptacion": tipo_adaptacion,
        "accion_rutina": accion
    }
```

```python
def evaluar_usuario(
    edad,
    peso,
    objetivo,
    dias,
    tiempo,
    experiencia,
    lugar,
    zona_lesion,
    estado_lesion
):
    dificultad = predecir_dificultad(
        edad,
        peso,
        objetivo,
        dias,
        tiempo,
        experiencia,
        lugar,
        zona_lesion if zona_lesion != "nada" else "nada"
    )

    if estado_lesion == "actual":
        requiere_adaptacion = "si"
        tipo_adaptacion = "especial"
        accion = f"limitar ejercicios de {zona_lesion} y requerir revision profesional"

    elif estado_lesion == "pasada":
        requiere_adaptacion = "si"
        tipo_adaptacion = "preventiva"
        accion = f"adaptar ejercicios para cuidar {zona_lesion}"

    else:
        requiere_adaptacion = "no"
        tipo_adaptacion = "ninguna"
        accion = "rutina normal"

    return {
        "dificultad": dificultad,
        "requiere_adaptacion": requiere_adaptacion,
        "tipo_adaptacion": tipo_adaptacion,
        "accion_rutina": accion
    }
```

```python
def obtener_prioridades_objetivo(objetivo):
    if objetivo == "ganar fuerza":
        return ["pecho", "espalda", "piernas", "hombros", "biceps", "triceps", "core"]
    elif objetivo == "mejorar resistencia":
        return ["cardio", "piernas", "core", "espalda", "pecho", "hombros", "biceps", "triceps"]
    elif objetivo == "bajar peso":
        return ["cardio", "piernas", "core", "pecho", "espalda", "hombros", "biceps", "triceps"]
    else:
        return ["pecho", "espalda", "piernas", "hombros", "core", "biceps", "triceps", "cardio"]
```

```python
def obtener_division(dias, objetivo):
    if objetivo == "mejorar resistencia":
        grupos = ["cardio", "piernas", "core", "espalda", "pecho", "hombros"]
    elif objetivo == "bajar peso":
        grupos = ["cardio", "piernas", "core", "pecho", "espalda", "hombros"]
    else:
        grupos = ["pecho", "espalda", "piernas", "hombros", "biceps", "triceps", "core"]

    division = {i + 1: [] for i in range(dias)}

    for i, grupo in enumerate(grupos):
        dia = (i % dias) + 1
        division[dia].append(grupo)

    return division
```

```python
def generar_plan_semanal(
    dias,
    lugar,
    dificultad,
    objetivo,
    zona_lesion="nada",
    cantidad_por_dia=3
):
    division = obtener_division(dias, objetivo)
    prioridades = obtener_prioridades_objetivo(objetivo)

    plan = {}
    ejercicios_usados = set()

    def filtrar_general(nivel, grupos=None):
        ejercicios = ejercicios_df[
            (ejercicios_df["lugar"] == lugar) &
            (ejercicios_df["dificultad"] == nivel)
        ].copy()

        if grupos is not None:
            ejercicios = ejercicios[
                ejercicios["grupo_muscular"].isin(grupos)
            ].copy()

        if zona_lesion != "nada" and not ejercicios.empty:
            mascara = ejercicios["restricciones"].apply(
                lambda restricciones: zona_lesion not in restricciones
            )
            ejercicios = ejercicios.loc[mascara].copy()

        if not ejercicios.empty:
            ejercicios = ejercicios[
                ~ejercicios["ejercicio"].isin(ejercicios_usados)
            ].copy()

        return ejercicios

    if dificultad == "Dificil":
        niveles = ["Dificil", "Medio", "Facil"]
    elif dificultad == "Medio":
        niveles = ["Medio", "Facil"]
    else:
        niveles = ["Facil"]

    for dia, grupos in division.items():
        candidatos_partes = []

        for nivel in niveles:
            parte = filtrar_general(nivel, grupos)
            if not parte.empty:
                candidatos_partes.append(parte)

        if candidatos_partes:
            candidatos = pd.concat(candidatos_partes).drop_duplicates(subset=["ejercicio"])
        else:
            candidatos = ejercicios_df.iloc[0:0].copy()

        if not candidatos.empty:
            candidatos["prioridad"] = candidatos["grupo_muscular"].apply(
                lambda grupo: prioridades.index(grupo) if grupo in prioridades else 999
            )
            candidatos = candidatos.sort_values("prioridad")

            seleccionados = candidatos.drop_duplicates(
                subset=["grupo_muscular"]
            ).head(cantidad_por_dia)

            if len(seleccionados) < cantidad_por_dia:
                faltan = cantidad_por_dia - len(seleccionados)
                extras = candidatos[
                    ~candidatos["ejercicio"].isin(seleccionados["ejercicio"])
                ]
                seleccionados = pd.concat([seleccionados, extras.head(faltan)])
        else:
            seleccionados = candidatos.copy()

        if len(seleccionados) < cantidad_por_dia:
            faltan = cantidad_por_dia - len(seleccionados)
            extras_partes = []

            for nivel in niveles:
                parte = filtrar_general(nivel)
                if not parte.empty:
                    extras_partes.append(parte)

            if extras_partes:
                candidatos_extra = pd.concat(extras_partes).drop_duplicates(subset=["ejercicio"])
                if not seleccionados.empty:
                    candidatos_extra = candidatos_extra[
                        ~candidatos_extra["ejercicio"].isin(seleccionados["ejercicio"])
                    ]
                seleccionados = pd.concat([
                    seleccionados,
                    candidatos_extra.head(faltan)
                ])

        seleccionados = seleccionados.drop(columns=["prioridad"], errors="ignore")
        seleccionados = seleccionados.head(cantidad_por_dia)

        if not seleccionados.empty:
            ejercicios_usados.update(seleccionados["ejercicio"].tolist())

        plan[dia] = seleccionados

    return plan
```

```python
def parametros_entrenamiento(dificultad):
    if dificultad == "Facil":
        return {"series": 2, "repeticiones": "8-10"}
    elif dificultad == "Medio":
        return {"series": 3, "repeticiones": "10-12"}
    else:
        return {"series": 4, "repeticiones": "10-15"}
```

```python
def agregar_parametros_plan(plan_semanal, dificultad):
    parametros = parametros_entrenamiento(dificultad)
    plan_con_parametros = {}

    for dia, rutina in plan_semanal.items():
        rutina = rutina.copy()
        rutina["series"] = parametros["series"]
        rutina["repeticiones"] = parametros["repeticiones"]
        plan_con_parametros[dia] = rutina

    return plan_con_parametros
```

```python
def generar_plan_personalizado(
    edad,
    peso,
    objetivo,
    dias,
    tiempo,
    experiencia,
    lugar,
    zona_lesion="nada",
    estado_lesion="ninguna"
):
    evaluacion = evaluar_usuario(
        edad=edad,
        peso=peso,
        objetivo=objetivo,
        dias=dias,
        tiempo=tiempo,
        experiencia=experiencia,
        lugar=lugar,
        zona_lesion=zona_lesion,
        estado_lesion=estado_lesion
    )

    cantidad = cantidad_ejercicios_segun_tiempo(tiempo)

    plan = generar_plan_semanal(
        dias=dias,
        lugar=lugar,
        dificultad=evaluacion["dificultad"],
        objetivo=objetivo,
        zona_lesion=zona_lesion,
        cantidad_por_dia=cantidad
    )

    plan = agregar_parametros_plan(plan, evaluacion["dificultad"])

    return {
        "evaluacion": evaluacion,
        "cantidad_ejercicios_por_dia": cantidad,
        "plan_semanal": plan
    }
```

```python
def generar_recomendaciones_alimentarias(
    objetivo,
    vegetariano="no",
    restricciones_usuario=None
):
    if restricciones_usuario is None:
        restricciones_usuario = []

    opciones = comidas_df[
        comidas_df["objetivo"].isin([objetivo, "general"])
    ].copy()

    if vegetariano == "si":
        opciones = opciones[
            opciones["vegetariano"] == "si"
        ]

    opciones = opciones[
        opciones["restricciones"].apply(
            lambda restricciones_comida:
            not any(
                restriccion in restricciones_comida
                for restriccion in restricciones_usuario
            )
        )
    ]

    return opciones
```

```python
def armar_plan_alimentario(recomendaciones):
    plan = {}

    tipos = ["desayuno", "almuerzo", "merienda", "cena"]

    for tipo in tipos:
        opciones = recomendaciones[
            recomendaciones["tipo"] == tipo
        ]

        if len(opciones) > 0:
            comida_elegida = opciones.sample(1).iloc[0]["comida"]
            plan[tipo] = comida_elegida
        else:
            plan[tipo] = "sin opcion disponible"

    return plan
```

```python
def generar_plan_alimentario_semanal(recomendaciones, dias=3):
    plan_semanal = {}
    comidas_usadas = set()

    tipos = ["desayuno", "almuerzo", "merienda", "cena"]

    for dia in range(1, dias + 1):
        plan_dia = {}

        for tipo in tipos:
            opciones = recomendaciones[
                recomendaciones["tipo"] == tipo
            ]

            opciones_no_usadas = opciones[
                ~opciones["comida"].isin(comidas_usadas)
            ]

            if len(opciones_no_usadas) > 0:
                comida = opciones_no_usadas.sample(1).iloc[0]["comida"]

            elif len(opciones) > 0:
                comida = opciones.sample(1).iloc[0]["comida"]

            else:
                comida = "sin opcion disponible"

            plan_dia[tipo] = comida

            if comida != "sin opcion disponible":
                comidas_usadas.add(comida)

        plan_semanal[dia] = plan_dia

    return plan_semanal
```

```python
plan_alimentario_3_dias = generar_plan_alimentario_semanal(
    recomendaciones,
    dias=3
)

plan_alimentario_3_dias
```
