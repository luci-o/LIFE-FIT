```python
def evaluar_usuario(
    edad,
    peso,
    objetivo,
    dias,
    tiempo,
    experiencia,
    lugar,
    zona_lesion,
    estado_lesion
):
    dificultad = predecir_dificultad(
        edad,
        peso,
        objetivo,
        dias,
        tiempo,
        experiencia,
        lugar,
        zona_lesion if zona_lesion != "nada" else "nada"
    )

    if estado_lesion == "actual":
        requiere_adaptacion = "si"
        tipo_adaptacion = "especial"
        accion = f"limitar ejercicios de {zona_lesion} y requerir revision profesional"

    elif estado_lesion == "pasada":
        requiere_adaptacion = "si"
        tipo_adaptacion = "preventiva"
        accion = f"adaptar ejercicios para cuidar {zona_lesion}"

    else:
        requiere_adaptacion = "no"
        tipo_adaptacion = "ninguna"
        accion = "rutina normal"

    return {
        "dificultad": dificultad,
        "requiere_adaptacion": requiere_adaptacion,
        "tipo_adaptacion": tipo_adaptacion,
        "accion_rutina": accion
    }
```

```python
def obtener_division(dias, objetivo):
    if objetivo == "mejorar resistencia":
        grupos = [
            "cardio",
            "piernas",
            "core",
            "espalda",
            "pecho",
            "hombros"
        ]

    elif objetivo == "bajar peso":
        grupos = [
            "cardio",
            "piernas",
            "core",
            "espalda",
            "pecho",
            "hombros"
        ]

    else:
        grupos = [
            "pecho",
            "espalda",
            "piernas",
            "hombros",
            "biceps",
            "triceps",
            "core"
        ]

    division = {
        i + 1: []
        for i in range(dias)
    }

    for i, grupo in enumerate(grupos):
        dia = (i % dias) + 1
        division[dia].append(grupo)

    return division
```

```python
def parametros_entrenamiento(dificultad):
    if dificultad == "Facil":
        return {
            "series": 2,
            "repeticiones": "8-10"
        }

    elif dificultad == "Medio":
        return {
            "series": 3,
            "repeticiones": "10-12"
        }

    else:
        return {
            "series": 4,
            "repeticiones": "10-15"
        }


def agregar_parametros_plan(
    plan_semanal,
    dificultad
):
    parametros = parametros_entrenamiento(
        dificultad
    )

    plan_con_parametros = {}

    for dia, rutina in plan_semanal.items():
        rutina = rutina.copy()

        rutina["series"] = parametros[
            "series"
        ]

        rutina["repeticiones"] = parametros[
            "repeticiones"
        ]

        plan_con_parametros[dia] = rutina

    return plan_con_parametros
```

```python
def generar_plan_semanal(
    dias,
    lugar,
    dificultad,
    objetivo,
    zona_lesion="nada",
    cantidad_por_dia=3
):
    division = obtener_division(
        dias,
        objetivo
    )

    plan = {}
    ejercicios_usados = set()

    if dificultad == "Dificil":
        niveles = [
            "Dificil",
            "Medio",
            "Facil"
        ]

    elif dificultad == "Medio":
        niveles = [
            "Medio",
            "Facil"
        ]

    else:
        niveles = [
            "Facil"
        ]

    def filtrar_ejercicios(grupos):
        ejercicios = ejercicios_df[
            (ejercicios_df["lugar"] == lugar) &
            (
                ejercicios_df[
                    "grupo_muscular"
                ].isin(grupos)
            ) &
            (
                ejercicios_df[
                    "dificultad"
                ].isin(niveles)
            )
        ].copy()

        if zona_lesion != "nada":
            ejercicios = ejercicios[
                ejercicios[
                    "restricciones"
                ].apply(
                    lambda restricciones:
                    zona_lesion
                    not in restricciones
                )
            ].copy()

        ejercicios["ya_usado"] = ejercicios[
            "ejercicio"
        ].apply(
            lambda ejercicio:
            1
            if ejercicio in ejercicios_usados
            else 0
        )

        ejercicios[
            "prioridad_dificultad"
        ] = ejercicios[
            "dificultad"
        ].apply(
            lambda nivel:
            niveles.index(nivel)
            if nivel in niveles
            else 999
        )

        ejercicios["azar"] = np.random.random(
            len(ejercicios)
        )

        ejercicios = ejercicios.sort_values(
            [
                "ya_usado",
                "prioridad_dificultad",
                "azar"
            ]
        )

        return ejercicios

    for dia, grupos_dia in division.items():
        candidatos = filtrar_ejercicios(
            grupos_dia
        )

        if candidatos.empty:
            plan[dia] = ejercicios_df.iloc[
                0:0
            ].copy()
            continue

        seleccionados = []
        nombres_seleccionados = set()

        while len(seleccionados) < cantidad_por_dia:
            agregado = False

            for grupo in grupos_dia:
                opciones = candidatos[
                    candidatos[
                        "grupo_muscular"
                    ] == grupo
                ]

                opciones = opciones[
                    ~opciones[
                        "ejercicio"
                    ].isin(
                        nombres_seleccionados
                    )
                ]

                if not opciones.empty:
                    ejercicio = opciones.iloc[0]

                    seleccionados.append(
                        ejercicio.to_dict()
                    )

                    nombres_seleccionados.add(
                        ejercicio["ejercicio"]
                    )

                    agregado = True

                if len(seleccionados) >= cantidad_por_dia:
                    break

            if not agregado:
                break

        seleccionados_df = pd.DataFrame(
            seleccionados
        )

        if not seleccionados_df.empty:
            seleccionados_df = (
                seleccionados_df.drop(
                    columns=[
                        "ya_usado",
                        "prioridad_dificultad",
                        "azar"
                    ],
                    errors="ignore"
                )
            )

            ejercicios_usados.update(
                seleccionados_df[
                    "ejercicio"
                ].tolist()
            )

        plan[dia] = (
            seleccionados_df
            .reset_index(drop=True)
        )

    return plan
```

```python
def generar_plan_personalizado(
    edad,
    peso,
    objetivo,
    dias,
    tiempo,
    experiencia,
    lugar,
    zona_lesion="nada",
    estado_lesion="ninguna"
):
    evaluacion = evaluar_usuario(
        edad=edad,
        peso=peso,
        objetivo=objetivo,
        dias=dias,
        tiempo=tiempo,
        experiencia=experiencia,
        lugar=lugar,
        zona_lesion=zona_lesion,
        estado_lesion=estado_lesion
    )

    cantidad = cantidad_ejercicios_segun_tiempo(
        tiempo
    )

    plan = generar_plan_semanal(
        dias=dias,
        lugar=lugar,
        dificultad=evaluacion["dificultad"],
        objetivo=objetivo,
        zona_lesion=zona_lesion,
        cantidad_por_dia=cantidad
    )

    plan = agregar_parametros_plan(
        plan,
        evaluacion["dificultad"]
    )

    return {
        "evaluacion": evaluacion,
        "cantidad_ejercicios_por_dia": cantidad,
        "plan_semanal": plan
    }
```

```python
def obtener_division(dias, objetivo):

    # GANAR FUERZA
    if objetivo == "ganar fuerza":

        if dias == 1:
            return {
                1: ["pecho", "espalda", "piernas", "hombros", "core"]
            }

        elif dias == 2:
            return {
                1: ["pecho", "espalda", "piernas"],
                2: ["hombros", "biceps", "triceps", "core"]
            }

        elif dias == 3:
            return {
                1: ["pecho", "piernas"],
                2: ["espalda", "biceps"],
                3: ["hombros", "triceps", "core"]
            }

        elif dias == 4:
            return {
                1: ["pecho", "triceps"],
                2: ["espalda", "biceps"],
                3: ["piernas", "core"],
                4: ["hombros", "pecho"]
            }

        elif dias == 5:
            return {
                1: ["pecho", "triceps"],
                2: ["espalda", "biceps"],
                3: ["piernas"],
                4: ["hombros", "core"],
                5: ["pecho", "espalda"]
            }

        elif dias == 6:
            return {
                1: ["pecho", "triceps"],
                2: ["espalda", "biceps"],
                3: ["piernas", "core"],
                4: ["pecho", "hombros"],
                5: ["espalda", "biceps"],
                6: ["piernas", "core"]
            }

        elif dias == 7:
            return {
                1: ["pecho", "triceps"],
                2: ["espalda", "biceps"],
                3: ["piernas", "core"],
                4: ["pecho", "hombros"],
                5: ["espalda", "biceps"],
                6: ["piernas", "core"],
                7: ["cardio", "core"]
            }

    # MEJORAR RESISTENCIA
    elif objetivo == "mejorar resistencia":

        if dias == 1:
            return {
                1: ["cardio", "piernas", "core", "espalda", "pecho"]
            }

        elif dias == 2:
            return {
                1: ["cardio", "piernas", "core"],
                2: ["cardio", "pecho", "espalda"]
            }

        elif dias == 3:
            return {
                1: ["cardio", "piernas"],
                2: ["espalda", "core", "cardio"],
                3: ["pecho", "hombros", "cardio"]
            }

        elif dias == 4:
            return {
                1: ["cardio", "piernas"],
                2: ["espalda", "core"],
                3: ["cardio", "pecho"],
                4: ["hombros", "piernas", "core"]
            }

        elif dias == 5:
            return {
                1: ["cardio", "piernas"],
                2: ["espalda", "core"],
                3: ["cardio", "pecho"],
                4: ["hombros", "piernas"],
                5: ["cardio", "core"]
            }

        elif dias == 6:
            return {
                1: ["cardio", "piernas"],
                2: ["espalda", "core"],
                3: ["cardio", "pecho"],
                4: ["piernas", "hombros"],
                5: ["cardio", "espalda"],
                6: ["core", "pecho"]
            }

        elif dias == 7:
            return {
                1: ["cardio", "piernas"],
                2: ["espalda", "core"],
                3: ["cardio", "pecho"],
                4: ["piernas", "hombros"],
                5: ["cardio", "espalda"],
                6: ["core", "pecho"],
                7: ["cardio", "core"]
            }

    # BAJAR PESO
    elif objetivo == "bajar peso":

        if dias == 1:
            return {
                1: ["cardio", "piernas", "core", "pecho", "espalda"]
            }

        elif dias == 2:
            return {
                1: ["cardio", "piernas", "pecho"],
                2: ["cardio", "espalda", "core"]
            }

        elif dias == 3:
            return {
                1: ["cardio", "piernas"],
                2: ["pecho", "espalda", "core"],
                3: ["cardio", "hombros", "piernas"]
            }

        elif dias == 4:
            return {
                1: ["cardio", "piernas"],
                2: ["pecho", "espalda"],
                3: ["cardio", "core"],
                4: ["hombros", "piernas"]
            }

        elif dias == 5:
            return {
                1: ["cardio", "piernas"],
                2: ["pecho", "espalda"],
                3: ["cardio", "core"],
                4: ["hombros", "piernas"],
                5: ["cardio", "pecho", "espalda"]
            }

        elif dias == 6:
            return {
                1: ["cardio", "piernas"],
                2: ["pecho", "espalda"],
                3: ["cardio", "core"],
                4: ["piernas", "hombros"],
                5: ["cardio", "espalda"],
                6: ["pecho", "core"]
            }

        elif dias == 7:
            return {
                1: ["cardio", "piernas"],
                2: ["pecho", "espalda"],
                3: ["cardio", "core"],
                4: ["piernas", "hombros"],
                5: ["cardio", "espalda"],
                6: ["pecho", "core"],
                7: ["cardio", "piernas"]
            }

    return {}
```

```python
def generar_plan_semanal(
    dias,
    lugar,
    dificultad,
    objetivo,
    zona_lesion="nada",
    cantidad_por_dia=3
):
    division = obtener_division(
        dias,
        objetivo
    )

    plan = {}
    ejercicios_usados = set()

    if dificultad == "Dificil":
        niveles = [
            "Dificil",
            "Medio",
            "Facil"
        ]

    elif dificultad == "Medio":
        niveles = [
            "Medio",
            "Facil"
        ]

    else:
        niveles = [
            "Facil"
        ]

    def filtrar_ejercicios(grupos):

        ejercicios = ejercicios_df[
            (ejercicios_df["lugar"] == lugar) &
            (
                ejercicios_df[
                    "grupo_muscular"
                ].isin(grupos)
            ) &
            (
                ejercicios_df[
                    "dificultad"
                ].isin(niveles)
            )
        ].copy()

        if zona_lesion != "nada":

            ejercicios = ejercicios[
                ejercicios[
                    "restricciones"
                ].apply(
                    lambda restricciones:
                    zona_lesion
                    not in restricciones
                )
            ].copy()

        ejercicios["ya_usado"] = ejercicios[
            "ejercicio"
        ].apply(
            lambda ejercicio:
            1
            if ejercicio in ejercicios_usados
            else 0
        )

        ejercicios[
            "prioridad_dificultad"
        ] = ejercicios[
            "dificultad"
        ].apply(
            lambda nivel:
            niveles.index(nivel)
            if nivel in niveles
            else 999
        )

        # Agrega variedad aleatoria
        ejercicios["azar"] = (
            np.random.random(
                len(ejercicios)
            )
        )

        ejercicios = ejercicios.sort_values(
            [
                "ya_usado",
                "prioridad_dificultad",
                "azar"
            ]
        )

        return ejercicios

    for dia, grupos_dia in division.items():

        candidatos = filtrar_ejercicios(
            grupos_dia
        )

        if candidatos.empty:

            plan[dia] = ejercicios_df.iloc[
                0:0
            ].copy()

            continue

        seleccionados = []
        nombres_seleccionados = set()

        while (
            len(seleccionados)
            < cantidad_por_dia
        ):

            agregado = False

            for grupo in grupos_dia:

                opciones = candidatos[
                    candidatos[
                        "grupo_muscular"
                    ] == grupo
                ]

                opciones = opciones[
                    ~opciones[
                        "ejercicio"
                    ].isin(
                        nombres_seleccionados
                    )
                ]

                if not opciones.empty:

                    ejercicio = opciones.iloc[0]

                    seleccionados.append(
                        ejercicio.to_dict()
                    )

                    nombres_seleccionados.add(
                        ejercicio["ejercicio"]
                    )

                    agregado = True

                if (
                    len(seleccionados)
                    >= cantidad_por_dia
                ):
                    break

            if not agregado:
                break

        seleccionados_df = pd.DataFrame(
            seleccionados
        )

        if not seleccionados_df.empty:

            seleccionados_df = (
                seleccionados_df.drop(
                    columns=[
                        "ya_usado",
                        "prioridad_dificultad",
                        "azar"
                    ],
                    errors="ignore"
                )
            )

            ejercicios_usados.update(
                seleccionados_df[
                    "ejercicio"
                ].tolist()
            )

        plan[dia] = (
            seleccionados_df
            .reset_index(drop=True)
        )

    return plan
```

```python
def parametros_entrenamiento(dificultad):
    if dificultad == "Facil":
        return {"series": 2, "repeticiones": "8-10"}
    elif dificultad == "Medio":
        return {"series": 3, "repeticiones": "10-12"}
    else:
        return {"series": 4, "repeticiones": "10-15"}
```

```python
def agregar_parametros_plan(plan_semanal, dificultad):
    parametros = parametros_entrenamiento(dificultad)
    plan_con_parametros = {}

    for dia, rutina in plan_semanal.items():
        rutina = rutina.copy()
        rutina["series"] = parametros["series"]
        rutina["repeticiones"] = parametros["repeticiones"]
        plan_con_parametros[dia] = rutina

    return plan_con_parametros
```

```python
def generar_plan_personalizado(
    edad,
    peso,
    objetivo,
    dias,
    tiempo,
    experiencia,
    lugar,
    zona_lesion="nada",
    estado_lesion="ninguna"
):
    evaluacion = evaluar_usuario(
        edad=edad,
        peso=peso,
        objetivo=objetivo,
        dias=dias,
        tiempo=tiempo,
        experiencia=experiencia,
        lugar=lugar,
        zona_lesion=zona_lesion,
        estado_lesion=estado_lesion
    )

    cantidad = cantidad_ejercicios_segun_tiempo(tiempo)

    plan = generar_plan_semanal(
        dias=dias,
        lugar=lugar,
        dificultad=evaluacion["dificultad"],
        objetivo=objetivo,
        zona_lesion=zona_lesion,
        cantidad_por_dia=cantidad
    )

    plan = agregar_parametros_plan(plan, evaluacion["dificultad"])

    return {
        "evaluacion": evaluacion,
        "cantidad_ejercicios_por_dia": cantidad,
        "plan_semanal": plan
    }
```

```python
def generar_plan_alimentario_semanal(recomendaciones, dias=3):
    plan_semanal = {}
    comidas_usadas = set()

    tipos = ["desayuno", "almuerzo", "merienda", "cena"]

    for dia in range(1, dias + 1):
        plan_dia = {}

        for tipo in tipos:
            opciones = recomendaciones[
                recomendaciones["tipo"] == tipo
            ]

            opciones_no_usadas = opciones[
                ~opciones["comida"].isin(comidas_usadas)
            ]

            if len(opciones_no_usadas) > 0:
                comida = opciones_no_usadas.sample(1).iloc[0]["comida"]

            elif len(opciones) > 0:
                comida = opciones.sample(1).iloc[0]["comida"]

            else:
                comida = "sin opcion disponible"

            plan_dia[tipo] = comida

            if comida != "sin opcion disponible":
                comidas_usadas.add(comida)

        plan_semanal[dia] = plan_dia

    return plan_semanal
```

```python
def generar_resumen_progreso(progreso_df):
    promedio = progreso_df["cumplimiento"].mean()

    peso_inicial = progreso_df.iloc[0]["peso_kg"]
    peso_actual = progreso_df.iloc[-1]["peso_kg"]
    cambio_peso = peso_actual - peso_inicial

    nivel = evaluar_cumplimiento(promedio)
    decision = decidir_ajuste_plan(nivel)

    return {
        "peso_inicial": peso_inicial,
        "peso_actual": peso_actual,
        "cambio_peso": round(cambio_peso, 2),
        "cumplimiento_promedio": round(promedio, 2),
        "nivel_cumplimiento": nivel,
        "decision": decision
    }
```

```python
def detectar_logros(progreso_df):
    logros = []

    promedio = progreso_df["cumplimiento"].mean()
    semanas = len(progreso_df)

    if semanas >= 1:
        logros.append("Primera semana registrada")

    if promedio >= 80:
        logros.append("Constancia alta")

    if (progreso_df["cumplimiento"] == 100).any():
        logros.append("Semana perfecta")

    if semanas >= 4:
        logros.append("Un mes de seguimiento")

    return logros
```

```python
import json
import numpy as np

def convertir_a_json(salida_final):
    salida_json = {
        "entrenamiento": {
            "evaluacion": salida_final["entrenamiento"]["evaluacion"],
            "cantidad_ejercicios_por_dia":
                salida_final["entrenamiento"]["cantidad_ejercicios_por_dia"],
            "plan_semanal": {}
        },
        "alimentacion": salida_final["alimentacion"],
        "progreso": salida_final["progreso"],
        "logros": salida_final["logros"]
    }

    for dia, rutina in salida_final["entrenamiento"]["plan_semanal"].items():
        salida_json["entrenamiento"]["plan_semanal"][dia] = (
            rutina.to_dict(orient="records")
        )

    def convertir_tipos(obj):
        if isinstance(obj, dict):
            return {
                clave: convertir_tipos(valor)
                for clave, valor in obj.items()
            }

        elif isinstance(obj, list):
            return [convertir_tipos(valor) for valor in obj]

        elif isinstance(obj, np.bool_):
            return bool(obj)

        elif isinstance(obj, np.integer):
            return int(obj)

        elif isinstance(obj, np.floating):
            return float(obj)

        else:
            return obj

    return convertir_tipos(salida_json)
```

```python
def generar_recomendaciones_alimentarias(
    objetivo,
    vegetariano="no",
    restricciones_usuario=None
):
    if restricciones_usuario is None:
        restricciones_usuario = []

    opciones = comidas_df.copy()

    
    if vegetariano == "si":
        opciones = opciones[
            opciones["vegetariano"] == "si"
        ]

    
    opciones = opciones[
        opciones["restricciones"].apply(
            lambda restricciones_comida:
            not any(
                restriccion in restricciones_comida
                for restriccion in restricciones_usuario
            )
        )
    ]

   
    opciones["prioridad_objetivo"] = opciones["objetivo"].apply(
        lambda x: 0 if x == objetivo else 1 if x == "general" else 2
    )

    opciones = opciones.sort_values(
        "prioridad_objetivo"
    ).reset_index(drop=True)

    return opciones
```

```python
def obtener_ingrediente_principal(comida):
    comida = comida.lower()

    equivalencias = {
        "pollo": ["pollo"],
        "huevo": ["huevo", "omelette", "tortilla"],
        "arroz": ["arroz"],
        "pasta": ["pasta"],
        "lentejas": ["lenteja", "lentejas"],
        "garbanzos": ["garbanzo", "garbanzos", "hummus"],
        "avena": ["avena", "panqueques de avena"],
        "yogur": ["yogur"],
        "atun": ["atun", "atún"],
        "fruta": ["fruta", "banana", "manzana"],
        "verduras": ["verdura", "verduras"]
    }

    for ingrediente, palabras in equivalencias.items():
        for palabra in palabras:
            if palabra in comida:
                return ingrediente

    return "otro"
```

```python
def generar_plan_alimentario_semanal(recomendaciones, dias=3):
    plan_semanal = {}

    comidas_usadas = set()
    contador_ingredientes = {}

    max_repeticiones_ingrediente = 2

    tipos = [
        "desayuno",
        "almuerzo",
        "merienda",
        "cena"
    ]

    for dia in range(1, dias + 1):
        plan_dia = {}
        ingredientes_del_dia = set()

        for tipo in tipos:
            opciones = recomendaciones[
                recomendaciones["tipo"] == tipo
            ].copy()

            if opciones.empty:
                plan_dia[tipo] = "sin opcion disponible"
                continue

            opciones["ingrediente_principal"] = opciones[
                "comida"
            ].apply(
                obtener_ingrediente_principal
            )

            # No repetir exactamente la misma comida
            opciones_no_usadas = opciones[
                ~opciones["comida"].isin(comidas_usadas)
            ].copy()

            if not opciones_no_usadas.empty:
                opciones = opciones_no_usadas

            # Evitar repetir ingrediente dentro del mismo día
            opciones_dia = opciones[
                ~opciones["ingrediente_principal"].isin(
                    ingredientes_del_dia
                )
            ].copy()

            if not opciones_dia.empty:
                opciones = opciones_dia

            # Evitar ingredientes que ya llegaron al máximo semanal
            opciones_limite = opciones[
                opciones["ingrediente_principal"].apply(
                    lambda ingrediente:
                    contador_ingredientes.get(
                        ingrediente,
                        0
                    ) < max_repeticiones_ingrediente
                )
            ].copy()

            if not opciones_limite.empty:
                opciones = opciones_limite

            # Preferir ingredientes menos usados
            opciones["veces_usado"] = opciones[
                "ingrediente_principal"
            ].apply(
                lambda ingrediente:
                contador_ingredientes.get(
                    ingrediente,
                    0
                )
            )

            minimo_usos = opciones[
                "veces_usado"
            ].min()

            opciones = opciones[
                opciones["veces_usado"] == minimo_usos
            ].copy()

            # Prioridad según objetivo
            if "prioridad_objetivo" in opciones.columns:
                mejor_prioridad = opciones[
                    "prioridad_objetivo"
                ].min()

                opciones = opciones[
                    opciones["prioridad_objetivo"]
                    == mejor_prioridad
                ]

            comida_elegida = opciones.sample(
                1
            ).iloc[0]

            comida = comida_elegida["comida"]
            ingrediente = comida_elegida[
                "ingrediente_principal"
            ]

            plan_dia[tipo] = comida

            comidas_usadas.add(comida)
            ingredientes_del_dia.add(
                ingrediente
            )

            contador_ingredientes[ingrediente] = (
                contador_ingredientes.get(
                    ingrediente,
                    0
                ) + 1
            )

        plan_semanal[dia] = plan_dia

    return plan_semanal
```

```python
def validar_datos_usuario(usuario):
    errores = []

    campos_obligatorios = [
        "edad",
        "peso",
        "objetivo",
        "dias",
        "tiempo",
        "experiencia",
        "lugar",
        "zona_lesion",
        "estado_lesion",
        "vegetariano",
        "restricciones_alimentarias"
    ]

    for campo in campos_obligatorios:
        if campo not in usuario:
            errores.append(f"Falta el campo: {campo}")

    if errores:
        return {
            "valido": False,
            "errores": errores
        }

    if usuario["edad"] <= 0:
        errores.append("La edad debe ser mayor a 0")

    if usuario["peso"] <= 0:
        errores.append("El peso debe ser mayor a 0")

    if usuario["dias"] < 1 or usuario["dias"] > 7:
        errores.append(
            "Los días disponibles deben estar entre 1 y 7"
        )

    if usuario["tiempo"] <= 0:
        errores.append(
            "El tiempo disponible debe ser mayor a 0"
        )

    objetivos_validos = [
        "ganar fuerza",
        "mejorar resistencia",
        "bajar peso"
    ]

    if usuario["objetivo"] not in objetivos_validos:
        errores.append("Objetivo no válido")

    experiencias_validas = [
        "principiante",
        "intermedio",
        "avanzado"
    ]

    if usuario["experiencia"] not in experiencias_validas:
        errores.append("Experiencia no válida")

    lugares_validos = [
        "gym",
        "hogar",
        "aire libre"
    ]

    if usuario["lugar"] not in lugares_validos:
        errores.append("Lugar de entrenamiento no válido")

    return {
        "valido": len(errores) == 0,
        "errores": errores
    }
```

```python
def generar_resultado_usuario(usuario):
    validacion = validar_datos_usuario(usuario)

    if not validacion["valido"]:
        return {
            "errores": validacion["errores"],
            "advertencias": [],
            "plan": None
        }

    advertencias = revisar_usuario(usuario)

    plan = generar_plan_completo_usuario(
        edad=usuario["edad"],
        peso=usuario["peso"],
        objetivo=usuario["objetivo"],
        dias=usuario["dias"],
        tiempo=usuario["tiempo"],
        experiencia=usuario["experiencia"],
        lugar=usuario["lugar"],
        zona_lesion=usuario["zona_lesion"],
        estado_lesion=usuario["estado_lesion"],
        vegetariano=usuario["vegetariano"],
        restricciones_alimentarias=usuario["restricciones_alimentarias"]
    )

    return {
        "errores": [],
        "advertencias": advertencias,
        "plan": plan
    }
```

```python
def asignar_duracion_cardio(rutina, tiempo_total):
    rutina = rutina.copy()

    cantidad_cardio = (
        rutina["grupo_muscular"] == "cardio"
    ).sum()

    if cantidad_cardio == 0:
        rutina["duracion"] = "-"
        return rutina

    tiempo_cardio_total = tiempo_total * 0.5
    minutos_por_cardio = max(
        5,
        int(tiempo_cardio_total / cantidad_cardio)
    )

    rutina["duracion"] = rutina["grupo_muscular"].apply(
        lambda grupo:
        f"{minutos_por_cardio} min"
        if grupo == "cardio"
        else "-"
    )

    return rutina
```

```python
def agregar_duracion_cardio_plan(plan_semanal, tiempo):
    plan_nuevo = {}

    for dia, rutina in plan_semanal.items():
        plan_nuevo[dia] = asignar_duracion_cardio(
            rutina,
            tiempo
        )

    return plan_nuevo
```

```python
def estimar_duracion_rutina(rutina):
    minutos_totales = 0

    for _, ejercicio in rutina.iterrows():

        if ejercicio["grupo_muscular"] == "cardio":
            duracion = ejercicio.get("duracion", "-")

            if isinstance(duracion, str) and "min" in duracion:
                minutos_totales += int(
                    duracion.replace(" min", "")
                )

        else:
            series = ejercicio["series"]

            
            minutos_totales += series * 2

    return minutos_totales
```

```python
def evaluar_tiempo_rutina(rutina, tiempo_disponible):
    duracion_estimada = estimar_duracion_rutina(rutina)

    diferencia = tiempo_disponible - duracion_estimada

    if duracion_estimada > tiempo_disponible:
        estado = "se pasa del tiempo disponible"

    elif diferencia >= 20:
        estado = "queda bastante tiempo disponible"

    else:
        estado = "entra bien en el tiempo disponible"

    return {
        "tiempo_disponible": tiempo_disponible,
        "duracion_estimada": duracion_estimada,
        "diferencia": diferencia,
        "estado": estado
    }
```

```python
def validar_datos_usuario(usuario):
    errores = []

    campos_obligatorios = [
        "edad",
        "peso",
        "objetivo",
        "dias",
        "tiempo",
        "experiencia",
        "lugar",
        "zona_lesion",
        "estado_lesion",
        "vegetariano",
        "restricciones_alimentarias"
    ]

    for campo in campos_obligatorios:
        if campo not in usuario:
            errores.append(f"Falta el campo: {campo}")

    if errores:
        return {
            "valido": False,
            "errores": errores
        }

    if usuario["edad"] <= 0:
        errores.append("La edad debe ser mayor a 0")

    if usuario["peso"] <= 0:
        errores.append("El peso debe ser mayor a 0")

    if usuario["dias"] < 1 or usuario["dias"] > 7:
        errores.append(
            "Los días disponibles deben estar entre 1 y 7"
        )

    if usuario["tiempo"] <= 0:
        errores.append(
            "El tiempo disponible debe ser mayor a 0"
        )

    objetivos_validos = [
        "ganar fuerza",
        "mejorar resistencia",
        "bajar peso"
    ]

    if usuario["objetivo"] not in objetivos_validos:
        errores.append("Objetivo no válido")

    experiencias_validas = [
        "principiante",
        "intermedio",
        "avanzado"
    ]

    if usuario["experiencia"] not in experiencias_validas:
        errores.append("Experiencia no válida")

    lugares_validos = [
        "gym",
        "hogar",
        "aire libre"
    ]

    if usuario["lugar"] not in lugares_validos:
        errores.append("Lugar de entrenamiento no válido")

    return {
        "valido": len(errores) == 0,
        "errores": errores
    }
```

```python
def explicar_dificultad_usuario(usuario, dificultad):
    explicaciones = []

    explicaciones.append(
        f"El modelo estimó una dificultad {dificultad}."
    )

    explicaciones.append(
        f"Se tuvo en cuenta que la experiencia del usuario es "
        f"{usuario['experiencia']}."
    )

    explicaciones.append(
        f"Dispone de {usuario['tiempo']} minutos por sesión."
    )

    explicaciones.append(
        f"Puede entrenar {usuario['dias']} días por semana."
    )

    explicaciones.append(
        f"El objetivo seleccionado es {usuario['objetivo']}."
    )

    return explicaciones
```

```python
def verificar_restricciones_rutina(
    respuesta,
    zona_lesion
):
    conflictos = []

    if zona_lesion == "nada":
        return {
            "segura_segun_filtros": True,
            "conflictos": []
        }

    for dia, ejercicios in respuesta["entrenamiento"].items():
        for ejercicio in ejercicios:

            restricciones = ejercicio.get(
                "restricciones",
                []
            )

            if zona_lesion in restricciones:
                conflictos.append({
                    "dia": dia,
                    "ejercicio": ejercicio["ejercicio"],
                    "restriccion": zona_lesion
                })

    return {
        "segura_segun_filtros": len(conflictos) == 0,
        "conflictos": conflictos
    }
```

```python
def procesar_usuario(usuario):
    resultado = generar_resultado_usuario_completo(
        usuario
    )

    if resultado.get("errores"):
        return preparar_json_frontend(
            usuario,
            resultado
        )

    resultado = agregar_explicacion_resultado(
        usuario,
        resultado
    )

    resultado = agregar_explicacion_alimentaria(
        usuario,
        resultado
    )

    salida_json = preparar_json_frontend(
        usuario,
        resultado
    )

    verificacion_lesion = verificar_restricciones_rutina(
        salida_json,
        usuario["zona_lesion"]
    )

    salida_json["verificacion_lesion"] = (
        verificacion_lesion
    )

    dificultad = resultado[
        "plan"
    ]["entrenamiento"]["evaluacion"]["dificultad"]

    salida_json["explicacion_dificultad"] = (
        explicar_dificultad_usuario(
            usuario,
            dificultad
        )
    )

    salida_json = agregar_explicacion_ejercicios(
        salida_json
    )

    return salida_json
```

```python
def agregar_explicacion_ejercicios(respuesta):
    for dia, ejercicios in respuesta["entrenamiento"].items():
        for ejercicio in ejercicios:
            musculo = ejercicio.get(
                "musculo_principal",
                ejercicio.get("grupo_muscular", "grupo muscular")
            )

            dificultad = ejercicio.get(
                "dificultad",
                "la dificultad indicada"
            )

            ejercicio["explicacion"] = (
                f"Trabaja principalmente {musculo} "
                f"y fue seleccionado para una rutina "
                f"de dificultad {dificultad}."
            )

    return respuesta
```

```python
def verificar_restricciones_rutina(
    respuesta,
    zona_lesion
):
    conflictos = []

    if zona_lesion == "nada":
        return {
            "segura_segun_filtros": True,
            "conflictos": []
        }

    for dia, ejercicios in respuesta[
        "entrenamiento"
    ].items():

        for ejercicio in ejercicios:

            restricciones = ejercicio.get(
                "restricciones",
                []
            )

            if zona_lesion in restricciones:
                conflictos.append({
                    "dia": dia,
                    "ejercicio": ejercicio["ejercicio"],
                    "restriccion": zona_lesion
                })

    return {
        "segura_segun_filtros": len(conflictos) == 0,
        "conflictos": conflictos
    }
```

```python
def evaluar_semana_usuario(
    dias_planificados,
    dias_completados
):
    if dias_planificados <= 0:
        return {
            "cumplimiento": 0,
            "nivel": "sin datos"
        }

    cumplimiento = (
        dias_completados
        / dias_planificados
    ) * 100

    if cumplimiento >= 80:
        nivel = "alto"

    elif cumplimiento >= 50:
        nivel = "medio"

    else:
        nivel = "bajo"

    return {
        "cumplimiento": round(
            cumplimiento,
            1
        ),
        "nivel": nivel
    }
```

```python
def decidir_ajuste_semana(resultado_semana):
    nivel = resultado_semana["nivel"]

    if nivel == "alto":
        return {
            "accion": "mantener",
            "cambio_ejercicios": 0
        }

    elif nivel == "medio":
        return {
            "accion": "revisar",
            "cambio_ejercicios": -1
        }

    elif nivel == "bajo":
        return {
            "accion": "simplificar",
            "cambio_ejercicios": -2
        }

    return {
        "accion": "sin cambios",
        "cambio_ejercicios": 0
    }
```

```python
def calcular_cantidad_semana_siguiente(
    tiempo,
    ajuste_semana
):
    cantidad_base = cantidad_ejercicios_segun_tiempo(
        tiempo
    )

    cantidad_final = max(
        2,
        cantidad_base
        + ajuste_semana["cambio_ejercicios"]
    )

    return {
        "cantidad_base": cantidad_base,
        "cantidad_final": cantidad_final
    }
```

```python
def generar_semana_siguiente(
    usuario,
    resultado_semana
):
    ajuste = decidir_ajuste_semana(
        resultado_semana
    )

    cantidades = calcular_cantidad_semana_siguiente(
        usuario["tiempo"],
        ajuste
    )

    dificultad_actual = predecir_dificultad(
        usuario["edad"],
        usuario["peso"],
        usuario["objetivo"],
        usuario["dias"],
        usuario["tiempo"],
        usuario["experiencia"],
        usuario["lugar"],
        usuario["zona_lesion"]
    )

    dificultad_nueva = ajustar_dificultad_semana(
        dificultad_actual,
        resultado_semana
    )

    plan = generar_plan_semanal(
        dias=usuario["dias"],
        lugar=usuario["lugar"],
        dificultad=dificultad_nueva,
        objetivo=usuario["objetivo"],
        zona_lesion=usuario["zona_lesion"],
        cantidad_por_dia=cantidades["cantidad_final"]
    )

    plan = agregar_parametros_plan(
        plan,
        dificultad_nueva
    )

    return {
        "ajuste": ajuste,
        "cantidades": cantidades,
        "dificultad_anterior": dificultad_actual,
        "dificultad_nueva": dificultad_nueva,
        "plan_semanal": plan
    }
```

```python
def ajustar_dificultad_semana(
    dificultad_actual,
    resultado_semana
):
    nivel = resultado_semana["nivel"]

    niveles_dificultad = [
        "Facil",
        "Medio",
        "Dificil"
    ]

    if dificultad_actual not in niveles_dificultad:
        return dificultad_actual

    posicion = niveles_dificultad.index(
        dificultad_actual
    )

    if nivel == "bajo":
        nueva_posicion = max(
            0,
            posicion - 1
        )

        return niveles_dificultad[
            nueva_posicion
        ]

    return dificultad_actual
```

```python
def generar_mensaje_ajuste_semanal(resumen_ajuste):
    cumplimiento = resumen_ajuste["cumplimiento"]
    dificultad_anterior = resumen_ajuste["dificultad_anterior"]
    dificultad_nueva = resumen_ajuste["dificultad_nueva"]
    cantidad_anterior = resumen_ajuste["cantidad_anterior"]
    cantidad_nueva = resumen_ajuste["cantidad_nueva"]

    partes = [
        f"Tu cumplimiento semanal fue del {cumplimiento}%."
    ]

    if dificultad_anterior == dificultad_nueva:
        partes.append(
            f"La dificultad se mantiene en {dificultad_nueva}."
        )
    else:
        partes.append(
            f"La dificultad cambia de {dificultad_anterior} "
            f"a {dificultad_nueva}."
        )

    if cantidad_anterior == cantidad_nueva:
        partes.append(
            f"La cantidad de ejercicios se mantiene en "
            f"{cantidad_nueva} por día."
        )
    else:
        partes.append(
            f"La cantidad de ejercicios cambia de "
            f"{cantidad_anterior} a {cantidad_nueva} por día."
        )

    return " ".join(partes)
```

```python
def generar_resumen_ajuste_semanal(
    usuario,
    resultado_semana
):
    semana = generar_semana_siguiente(
        usuario,
        resultado_semana
    )

    resumen = {
        "cumplimiento": resultado_semana["cumplimiento"],
        "nivel": resultado_semana["nivel"],
        "accion": semana["ajuste"]["accion"],
        "dificultad_anterior": semana["dificultad_anterior"],
        "dificultad_nueva": semana["dificultad_nueva"],
        "cantidad_anterior": semana["cantidades"]["cantidad_base"],
        "cantidad_nueva": semana["cantidades"]["cantidad_final"]
    }

    resumen["mensaje"] = generar_mensaje_ajuste_semanal(
        resumen
    )

    return resumen
```

```python
def crear_nombre_archivo_imagen(nombre):
    nombre = nombre.lower()

    reemplazos = {
        "á": "a",
        "é": "e",
        "í": "i",
        "ó": "o",
        "ú": "u",
        "ñ": "n"
    }

    for original, nuevo in reemplazos.items():
        nombre = nombre.replace(original, nuevo)

    nombre = nombre.replace(" ", "_")

    return f"images/{nombre}.png"


ejercicios_df["imagen"] = ejercicios_df.apply(
    lambda fila:
        crear_nombre_archivo_imagen(
            fila["ejercicio"]
        )
        if (
            pd.isna(fila["imagen"]) or
            fila["imagen"] == "images/default.png"
        )
        else fila["imagen"],
    axis=1
)
```
